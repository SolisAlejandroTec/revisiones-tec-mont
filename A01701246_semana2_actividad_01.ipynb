*   **Pregunta 1.**

Busca y elimina todos los saltos de línea '\n' que se encuentran al final de cada comentario.

Una vez finalizado, imprime los primeros 10 comentarios del resultado obtenido.


```python
cleaned_docs = [doc.strip() for doc in docs]
```

```python
print(cleaned_docs[0:10])
```

*   **Pregunta 2.**  

Busca e imprime todas las palabras que terminan con dos o más signos de admiración seguidos, por ejemplo "!!!".

Debes imprimir tanto la palabra como la totalidad de signos de admiración que le siguen.

Indica cuántos resultados obtuviste.



```python
exclamation_matches = []
pattern = r'\b\w+!{2,}'

for doc in cleaned_docs:
    matches = re.findall(pattern, doc)
    if matches:
        exclamation_matches.extend(matches)

print("Palabras que terminan con dos o más signos de admiración:")
for match in exclamation_matches:
    print(match)
```

```python
print(f"Total de resultados obtenidos: {len(exclamation_matches)}")
```

*   **Pregunta 3.**  

Busca e imprime todas las palabras que están escritas totalmente en mayúsculas. Cada coincidencia debe ser una sola palabra.

Indica cuántas palabras encontraste.



```python
uppercase_words = []
pattern = r'\b[A-Z]+\b'

for doc in cleaned_docs:
    matches = re.findall(pattern, doc)
    if matches:
        uppercase_words.extend(matches)

print("Palabras en mayúsculas:")
for word in uppercase_words:
    print(word)
```

```python
print(f"\nTotal de palabras en mayúsculas: {len(uppercase_words)}")
```

*   **Pregunta 4.**  

Busca e imprime los comentarios en donde todos los caracteres alfabéticos (letras) están en mayúsculas.

Cada coincidencia encontrada debe ser todo el comentario/enunciado.

Indica cuántos resultados obtuviste.


```python
all_caps_comments = []

for doc in cleaned_docs:

    alphabetic_chars = re.sub(r'[^a-zA-Z]', '', doc)
    if alphabetic_chars and alphabetic_chars.isupper():
        all_caps_comments.append(doc)

print("Comentarios con todos los caracteres alfabéticos en mayúsculas:")
for comment in all_caps_comments:
    print(comment)
```

```python
print(f"\nTotal de comentarios encontrados: {len(all_caps_comments)}")
```

*   **Pregunta 5.**  

Busca e imprime todas las palabras que tengan una vocal acentuada, del tipo á, é, í, ó, ú.

Indica cuántos resultados obtuviste.

```python
accented_words = []
# Pattern to match words containing any of the accented vowels
pattern = r'\b\w*[áéíóúÁÉÍÓÚ]\w*\b'

for doc in cleaned_docs:
    matches = re.findall(pattern, doc, re.IGNORECASE)
    if matches:
        accented_words.extend(matches)

print("Palabras con vocales acentuadas:")
# Print unique words to avoid duplicates if a word appears multiple times
for word in sorted(list(set(accented_words))):
    print(word)
```

```python
print(f"\nTotal de palabras encontradas: {len(set(accented_words))}")
```

*   **Pregunta 6.**  

Busca e imprime todas las cantidades numéricas monetarias, enteras o con decimales, que inician con el símbolo $\$$.

Indica cuántos resultados obtuviste.

```python
monetary_amounts = []

pattern = r'\$\d+(?:\.\d+)?'

for doc in cleaned_docs:
    matches = re.findall(pattern, doc)
    if matches:
        monetary_amounts.extend(matches)

print("Cantidades numéricas monetarias encontradas:")
for amount in monetary_amounts:
    print(amount)
```

```python
print(f"\nTotal de resultados obtenidos: {len(monetary_amounts)}")
```

*   **Pregunta 7.**  

Busca e imprime todas las palabras que sean variantes de la palabra "love", sin importar si incluyen mayúsculas o minúsculas, o la manera en que esté conjugada o alguna otra variación que se haga con dicha palabra.

Indica cuántos resultados obtuviste.

```python
love_variants = []

pattern = r'\blov(?:e|ing|ed|es|s)?\b'

for doc in cleaned_docs:
    matches = re.findall(pattern, doc, re.IGNORECASE)
    if matches:
        love_variants.extend(matches)

print("Variantes de 'love' encontradas:")
for variant in love_variants:
    print(variant)
```

```python
print(f"\nTotal de resultados obtenidos: {len(love_variants)}")
```

*   **Pregunta 8.**  

Busca e imprime todas las palabras, variantes de "so" y "good", que tengan dos o más "o" en "so" y 3 o más "o" en good.

Indica cuántas encontraste.


```python
so_good_variants = []

pattern = r'\b(s[oO]{2,}|g[oO]{2,}d{2,})\b'

for doc in cleaned_docs:
    matches = re.findall(pattern, doc, re.IGNORECASE)
    if matches:
        so_good_variants.extend(matches)

print("Variantes de 'so' y 'good' encontradas:")

for variant in sorted(list(set(so_good_variants))):
    print(variant)
```

```python
print(f"\nTotal de resultados obtenidos: {len(set(so_good_variants))}")
```

*   **Pregunta 9.**  

Busca e imprime todas las palabras que tengan una longitud mayor estrictamente a 10 caracteres alfabéticos.

No se consideran los signos de puntuación o caracteres especiales en la longitud de estas cadenas, solo caracteres alfabéticos en mayúsculas o minúsculas.

Indica la cantidad de palabras encontradas.


```python
long_words = []

pattern = r'\b[a-zA-Z]+\b'

for doc in cleaned_docs:

    words_in_doc = re.findall(pattern, doc)

    for word in words_in_doc:
        if len(word) > 10:
            long_words.append(word)

print("Palabras con más de 10 caracteres alfabéticos:")
for word in long_words:
    print(word)
```

```python
print(f"\nTotal de palabras encontradas: {len(long_words)}")
```

*   **Pregunta 10.**  

Busca e imprime todas las palabras que inician con una letra mayúscula y terminan con una minúscula, pero que además no sea la primera palabra del comentario/string.

Indica la cantidad de resultados obtenidos.

```python
capitalized_words_not_first = []
pattern = r'\b([A-Z][a-z]+)\b'

for doc in cleaned_docs:

    for match in re.finditer(pattern, doc):

        if match.start() > 0:
            capitalized_words_not_first.append(match.group(1))

print("Palabras que inician con mayúscula, terminan con minúscula y no son la primera palabra:")
for word in capitalized_words_not_first:
    print(word)
```

```python
print(f"\nTotal de resultados obtenidos: {len(capitalized_words_not_first)}")
```

*   **Pregunta 11.**  

Busca e imprime la secuencia de dos o más palabras que están separadas por un guion, "-", sin que tengan espacios en blanco entre ellas.

Por ejemplo "Go-Kart" sería válido, pero "Go  -Kart" o "Go  -  Kart" no lo serían.

Indica la cantidad de resultados obtenidos.

```python
hyphenated_words = []

pattern = r'\b\w+-\w+(?:-\w+)*\b'

for doc in cleaned_docs:
    matches = re.findall(pattern, doc)
    if matches:
        hyphenated_words.extend(matches)

print("Secuencias de palabras separadas por un guion sin espacios:")
for word in hyphenated_words:
    print(word)
```

```python
print(f"\nTotal de resultados obtenidos: {len(hyphenated_words)}")
```

*   **Pregunta 12.**  

Busca e imprime todas las palabras que terminan en "ing" o "ed".

Indica la cantidad de palabras que encontraste de cada una.

```python
ing_words = []
ed_words = []

pattern_ing = r'\b\w+ing\b'
pattern_ed = r'\b\w+ed\b'

for doc in cleaned_docs:
    ing_matches = re.findall(pattern_ing, doc, re.IGNORECASE)
    if ing_matches:
        ing_words.extend(ing_matches)

    ed_matches = re.findall(pattern_ed, doc, re.IGNORECASE)
    if ed_matches:
        ed_words.extend(ed_matches)

print("Palabras que terminan en 'ing':")
for word in ing_words:
    print(word)
print(f"\nTotal de palabras terminadas en 'ing': {len(ing_words)}")

print("\nPalabras que terminan en 'ed':")
for word in ed_words:
    print(word)
```

```python
print(f"\nTotal de palabras terminadas en 'ed': {len(ed_words)}")
```

#**Parte 3. Proceso de limpieza.**

*   **Pregunta 13.**  

Ahora realiza un proceso de limpieza del corpus que incluya los siguientes procesos:

*   Solo se deben considerar caracteres alfabéticos. Es decir, se eliminan todos los signos de puntuación y caracteres especiales.
*   Todos los caracteres alfabéticos se transforman a minúsculas.
*   Se deben eliminar todos los espacios en blanco adicionales que se puedan encontrar en cada comentario.

Al finalizar dicho proceso de limpieza, imprime el resultado de los primeros 10 comentarios resultantes.
   




```python
fully_cleaned_docs = []

for doc in cleaned_docs:

    clean_doc = re.sub(r'[^a-zA-Z\s]', '', doc)

    clean_doc = clean_doc.lower()

    clean_doc = re.sub(r'\s+', ' ', clean_doc).strip()
    fully_cleaned_docs.append(clean_doc)

print("Primeros 10 comentarios después de la limpieza:")
for i, comment in enumerate(fully_cleaned_docs[:10]):
    print(f"{i+1}. {comment}")
```

*   **Pregunta 14.**  

Con el resultado de la limpieza obtenido en la pregunta anterior, realiza ahora un proceso de tokenización por palabras del corpus.

Es decir, al final de este proceso de tokenización, debes tener como resultado una lista de listas, donde cada comentario estará tokenizado por palabras.

Al terminar calcula el total de tokens obtenido en todo el corpus.

```python
tokenized_docs = []
total_tokens = 0

for doc in fully_cleaned_docs:
    tokens = doc.split()
    tokenized_docs.append(tokens)
    total_tokens += len(tokens)

print("Primeros 10 comentarios tokenizados:")
for i, tokens in enumerate(tokenized_docs[:10]):
    print(f"{i+1}. {tokens}")
```

```python
print(f"\nTotal de tokens en todo el corpus: {total_tokens}")
```

*   **Pregunta 15.**  

Finalmente, en este ejercicio definiremos nuestro conjunto de palabras "stopwords", las cuales deberás eliminar de todo el corpus.

Recuerda que ejemplos de stopwords son artículos, adverbios, conectivos, etcétera, que tienen frecuencias de aparición muy altas en cualquier documento, pero que no brindan mucho significado en cuanto al significado de un enunciado.

Con base a la lista de stopwords que se te proporciona, realiza un proceso de limpieza eliminando todas estas palabras del corpus obtenido en el ejercicio anterior.

Obtener cuántos tokens/palabras quedan finalmente en todo el corpus.

Obtener cuántos de estos tokens/palabras son diferentes, es decir, cuántos tokens únicos tendrá lo que llamaremos más adelante nuestro vocabulario.

```python
# Considera la siguiente lista como tu conjunto de stopwords:
mis_stopwords = ['i', 'me', 'my', 'myself', 'we', 'our', 'ours', 'ourselves', 'you', 'your', 'yours', 'he', 'him', 'his', 'himself', 'she', 'her', 'hers', 'herself', 'it', 'its', 'itself', 'they', 'them', 'their', 'theirs', 'themselves', 'what', 'which', 'who', 'whom', 'this', 'that', 'these', 'those', 'am', 'is', 'are', 'was', 'were', 'be', 'been', 'being', 'have', 'has', 'had', 'having', 'do', 'does', 'did', 'doing', 'a', 'an', 'the', 'and', 'but', 'if', 'or', 'because', 'as', 'until', 'while', 'of', 'at', 'by', 'for', 'with', 'about', 'against', 'between', 'into', 'through', 'during', 'before', 'after', 'above', 'below', 'to', 'from', 'up', 'down', 'in', 'out', 'on', 'over', 'under', 'again', 'further', 'then', 'once', 'here', 'there', 'when', 'where', 'why', 'how', 'all', 'any', 'both', 'each', 'few', 'more', 'most', 'other', 'some', 'such', 'only', 'own', 'same', 'so', 'than', 'too', 'very', 's', 't', 'can', 'will', 'just', 'should', 'now', 'll']
```

```python
filtered_tokenized_docs = []
for doc_tokens in tokenized_docs:
    filtered_doc = [token for token in doc_tokens if token not in mis_stopwords]
    filtered_tokenized_docs.append(filtered_doc)

# Calculate total tokens remaining
total_tokens_after_stopwords = sum(len(doc) for doc in filtered_tokenized_docs)

# Calculate unique tokens (vocabulary size)
vocabulary = set()
for doc_tokens in filtered_tokenized_docs:
    vocabulary.update(doc_tokens)
vocabulary_size = len(vocabulary)

print(f"\nTotal de tokens después de eliminar stopwords: {total_tokens_after_stopwords}")
print(f"Total de tokens únicos (tamaño del vocabulario): {vocabulary_size}")
```
