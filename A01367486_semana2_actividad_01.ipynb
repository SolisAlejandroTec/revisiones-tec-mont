*   **Pregunta 4.**

Cada coincidencia encontrada debe ser todo el comentario/enunciado.

Indica cuántos resultados obtuviste.

*   **Pregunta 8.**  

Busca e imprime todas las palabras, variantes de "so" y "good", que tengan dos o más "o" en "so" y 3 o más "o" en good.

Indica cuántas encontraste.


```python
pattern = r'\bso{2,}[!?,.:;]*|\bgo{3,}d[!?,.:;]*'

matches = []

for doc in docs:
    found = re.findall(pattern, doc, flags=re.IGNORECASE)
    matches.extend(found)
```

```python
for word in matches:
    print(word)

print("\nTotal encontradas:", len(matches))
```

*   **Pregunta 9.**  

Busca e imprime todas las palabras que tengan una longitud mayor estrictamente a 10 caracteres alfabéticos.

No se consideran los signos de puntuación o caracteres especiales en la longitud de estas cadenas, solo caracteres alfabéticos en mayúsculas o minúsculas.

Indica la cantidad de palabras encontradas.


```python
matches = []

for doc in docs:
    # Extrae solo palabras formadas por letras
    words = re.findall(r'[A-Za-z]+', doc)

    for word in words:
        if len(word) > 10:
            matches.append(word)
```

```python
for word in matches:
    print(word)

print("\nTotal encontradas:", len(matches))
```

*   **Pregunta 10.**  

Busca e imprime todas las palabras que inician con una letra mayúscula y terminan con una minúscula, pero que además no sea la primera palabra del comentario/string.

Indica la cantidad de resultados obtenidos.

```python
matches = []

for doc in docs:
    # Extrae solo palabras (sin signos)
    words = re.findall(r'[A-Za-z]+', doc)

    # Recorre desde la segunda palabra (índice 1)
    for word in words[1:]:
        if word[0].isupper() and word[-1].islower():
            matches.append(word)
```

```python
for word in matches:
    print(word)

print("\nTotal encontradas:", len(matches))
```

*   **Pregunta 11.**  

Busca e imprime la secuencia de dos o más palabras que están separadas por un guion, "-", sin que tengan espacios en blanco entre ellas.

Por ejemplo "Go-Kart" sería válido, pero "Go  -Kart" o "Go  -  Kart" no lo serían.

Indica la cantidad de resultados obtenidos.

```python
pattern = r'\b[A-Za-z]+(?:-[A-Za-z]+)+\b'

matches = []

for doc in docs:
    found = re.findall(pattern, doc)
    matches.extend(found)
```

```python
for seq in matches:
    print(seq)

print("\nTotal encontradas:", len(matches))
```

*   **Pregunta 12.**  

Busca e imprime todas las palabras que terminan en "ing" o "ed".

Indica la cantidad de palabras que encontraste de cada una.

```python
pattern = r'\b[A-Za-z]+(?:ing|ed)\b'

ing_words = []
ed_words = []

for doc in docs:
    found = re.findall(pattern, doc, flags=re.IGNORECASE)

    for word in found:
        if word.lower().endswith('ing'):
            ing_words.append(word)
        elif word.lower().endswith('ed'):
            ed_words.append(word)
```

```python
print("Palabras con 'ing':")
for word in ing_words:
    print(word)

print("\nPalabras con 'ed':")
for word in ed_words:
    print(word)

print("\nTotal 'ing':", len(ing_words))
print("\nTotal 'ed':", len(ed_words))
```

#**Parte 3. Proceso de limpieza.**

*   **Pregunta 13.**  

Ahora realiza un proceso de limpieza del corpus que incluya los siguientes procesos:

*   Solo se deben considerar caracteres alfabéticos. Es decir, se eliminan todos los signos de puntuación y caracteres especiales.
*   Todos los caracteres alfabéticos se transforman a minúsculas.
*   Se deben eliminar todos los espacios en blanco adicionales que se puedan encontrar en cada comentario.

Al finalizar dicho proceso de limpieza, imprime el resultado de los primeros 10 comentarios resultantes.
   




```python
clean_docs = []

for doc in docs:
    #only alph with spanish
    text = re.sub(r'[^A-Za-zÁÉÍÓÚÜÑáéíóúüñ\s]', '', doc)

    #lowercase
    text = text.lower()

    #remove space
    text = re.sub(r'\s+', ' ', text).strip()

    clean_docs.append(text)
```

```python
clean_docs[0:10]
```

*   **Pregunta 14.**  

Con el resultado de la limpieza obtenido en la pregunta anterior, realiza ahora un proceso de tokenización por palabras del corpus.

Es decir, al final de este proceso de tokenización, debes tener como resultado una lista de listas, donde cada comentario estará tokenizado por palabras.

Al terminar calcula el total de tokens obtenido en todo el corpus.

```python
tokenized_docs = []

for doc in clean_docs:
    tokens = doc.split()
    tokenized_docs.append(tokens)
```

```python
total_tokens = sum(len(tokens) for tokens in tokenized_docs)

print("Total de tokens en el corpus:", total_tokens)
```

*   **Pregunta 15.**

```python
filtered_docs = []

for tokens in tokenized_docs:
    filtered = [word for word in tokens if word not in mis_stopwords]
    filtered_docs.append(filtered)
```

```python
total_tokens = sum(len(tokens) for tokens in filtered_docs)
unique_tokens = set()

for tokens in filtered_docs:
    unique_tokens.update(tokens)

print("Total tokens:", total_tokens)
print("unique tokens:", len(unique_tokens))
```
