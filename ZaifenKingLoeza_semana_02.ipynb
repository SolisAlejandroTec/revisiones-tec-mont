#**Parte 3. Proceso de limpieza.**

*   **Pregunta 13.**  

Ahora realiza un proceso de limpieza del corpus que incluya los siguientes procesos:

*   Solo se deben considerar caracteres alfabéticos, incluyendo algún tipo de acentuación. Es decir, se eliminan todos los signos de puntuación y caracteres especiales.
*   Todos los caracteres alfabéticos se transforman a minúsculas.
*   Se deben eliminar todos los espacios en blanco adicionales que se puedan encontrar en cada comentario.

Al finalizar dicho proceso de limpieza, imprime el resultado de los primeros 10 comentarios resultantes.
   




```python
def limpiar_texto(s):
    # 1. Dejar solo letras (incluyendo acentos) y espacios
    s = re.sub(r'[^a-zA-ZáéíóúÁÉÍÓÚñÑ\s]', '', s)

    # 2. Convertir a minúsculas
    s = s.lower()

    # 3. Normalizar espacios (quitar extras)
    s =  ' '.join(s.split())

    return s
```

```python
doc_limpio = [limpiar_texto(s) for s in doc_clean]
```

```python
doc_limpio[0:10]
```
